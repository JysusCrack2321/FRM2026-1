## Introducción

En esta práctica de laboratorio se aborda el reconocimiento y la programación del robot móvil EV3, trabajando en trayectorias, acoples físicos, programación en lógica de bloques y manejo de sensores. Se desarrollan actividades que incluyen el movimiento controlado del robot en línea recta, la generación de trayectorias curvas y el uso de sensores (ultrasónico, infrarrojo, de contacto y giroscopio) para la toma de decisiones autónomas. Cada ejercicio es programado en el software LEGO MINDSTORMS Education EV3, ejecutado en el robot y documentado con mediciones, diagramas de flujo y registros audiovisuales, con el fin de analizar el comportamiento real del sistema.

## Punto 1: Movimiento en Línea Recta

**Objetivo:** Realizar desplazamientos rectilíneos controlados por tiempo y por encoder, y utilizar el sensor ultrasónico para detenerse ante un obstáculo.

**Procedimiento:**  

- Se programó el robot para moverse a velocidad constante durante 1 segundos y se midió la distancia recorrida.  
- Se varió el tiempo para verificar la proporcionalidad.  
- Se midió el radio de las ruedas y se calculó el avance por revolución.  
- Se investigó el encoder (pulsos por revolución) y se programó un movimiento de 8 vueltas exactas.  
- Finalmente, se implementó un programa que hace avanzar el robot hasta detectar un obstáculo a 10 cm con el sensor ultrasónico.

**Cálculos:**  

- Radio de la rueda: [2.65] cm  
- Avance por revolución: [16.65] cm  
- Distancia para 1 segundo: [31.5] cm

## Punto 2: Movimiento con Curva

**Objetivo:** Programar trayectorias curvas variando la velocidad relativa de las ruedas y generar una trayectoria en forma de “S”.

**Procedimiento:**  

- Se fijó la velocidad de una rueda a la mitad de la otra y se observó la trayectoria.  
- Se midió aproximadamente el radio de giro.  
- Finalmente se combinaron giros a derecha e izquierda para formar una “S”.

Para determinar cual es el radio de giro resultante se utilizo Traker obteniedo un valor de 30cm para una configuración de potencia en las ruedas del robot de 25/50%.

## Punto 3: Girosensor

**Objetivo:** Utilizar el giroscopio para realizar giros precisos de 45° y 135°.

**Procedimiento:**

- Se estudió el funcionamiento del sensor de giro (modo de medición de ángulo).  
- Se programó: avanzar en recta, girar 45°, avanzar de nuevo.  
- Se repitió para 135°.  

Utilizando la herramienta Traker se midio el angulo obtenido experimentalmente obteniendo un angulo de 51° este valor se acerca al programado pero debido al delay entre la lectura del angulo del giroscopio y la inercia rotacional del robot, el angulo final es mayor al programado.

## 7. Experimentación y dificultades
        
Durante el desarrollo del laboratorio se realizaron diversas pruebas experimentales que permitieron identificar fallas, validar hipótesis y mejorar la comprensión del comportamiento del robot móvil. A partir de estas pruebas se evidenciaron dificultades técnicas relacionadas principalmente con los sensores y la calibración del movimiento.,

### 1. Dificultad con el sensor infrarrojo

Una de las primeras dificultades surgió durante las pruebas con el sensor infrarrojo. Las mediciones obtenidas se encontraban fuera del rango esperado, lo que generó dudas sobre el correcto funcionamiento del sistema.,
Inicialmente se verificaron las conexiones entre el ladrillo inteligente (EV3) y el sensor, descartando fallas eléctricas o problemas de conexión del sensor. Al no detectarse anomalías en el cableado ni en la configuración del puerto, se procedió a realizar una prueba comparativa utilizando otro sensor infrarrojo.,
Al reemplazar el sensor y repetir las mediciones, se comprobó que las nuevas lecturas sí se encontraban dentro del rango adecuado. Esto permitió concluir que el inconveniente estaba asociado al primer sensor y no al sistema en general. Gracias a esta validación se pudo continuar y finalizar el laboratorio.,
Esta experiencia deja como aprendizaje metodológico que, ante mediciones inconsistentes, se pueden realizar pruebas cruzadas con otros componentes para verificar o descartar  errores en el componente (en este caso el sensor infrarrojo),

### 2. Dificultad con el girosensor

Otra dificultad se presentó al utilizar el girosensor para realizar un giro de 135 grados. Durante las primeras pruebas, el robot no alcanzaba el ángulo requerido, lo que indicaba un problema de calibración o de control en el programa.,
Para abordar esta situación, realizamos ajustes en el software de LEGO Mindstorms, particularmente en los parámetros de tiempo y ejecución del bloque de control. Se modificó el tiempo de activación del giro, reduciéndolo de 1 segundo a 0.01 segundos, lo que permitió un control más preciso y que el robot alcanzara el ángulo deseado.,
Este proceso evidenció la importancia de ajustar adecuadamente las variables de control, ya que pequeñas variaciones en el tiempo de ejecución pueden generar diferencias significativas en el resultado final del movimiento angular.,

### 3. Uso de herramientas de medición externas

El uso de herramientas adicionales fue fundamental para comprobar los resultados obtenidos según el punto de la guia de laboratorio. Por ejemplo el metro permitió medir con mayor precisión los desplazamientos lineales del robot y verificar si coincidían con lo solicitado en el laboratorio, tanto en trayectorias rectas como en movimientos curvos.,
Asimismo, la herramienta Tracker fue utilizada para analizar el movimiento del robot mediante registro visual. Esto permitió complementar y validar las mediciones del girosensor y del desplazamiento en curvas, ofreciendo una comparación entre los datos experimentales y el comportamiento real observado.

### 4. Dificultad con el sensor ultrasónico
Con el sensor ultrasónico, se tuvo una dificultad importante y fue que al momento de ejecutar la lógica en el software de lego, el robot seguía las instrucciones pero colisionaba con la pared utilizada como obstáculo, lo que impedía obtener mediciones concretas y confiables. 
Luego de rflexionar y compartir algunas hipótesis se llegó a la conclusión de que el robot se estrellaba por la inercia del movimiento del robot (agarre de las ruedas, tipo de suelo en las baldosas etc). Una vez tenido esto en cuenta se realizó una reducción desde el software del tiempo en que el sensor medía pasando de 1 segundo a 0.01 segundos.

### Conclusión experimental

Las dificultades encontradas durante el laboratorio no solo fueron resueltas, sino que fortalecieron la comprensión del funcionamiento de los sensores y del control de movimiento en un robot móvil. La experimentación, la validación cruzada de resultados obtenidos con otras herramientas externas de medición fueron claves para garantizar resultados coherentes y confiables.
